```python
# ==========================
# Imports
# ==========================

import tensorflow as tf
import os

from tensorflow.keras.models import load_model
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.callbacks import ReduceLROnPlateau, EarlyStopping
```

```python
# ==========================
# Unzip Dataset
# ==========================

!unzip merged_dataset.zip -d /content/merged_dataset
```

```python
# ==========================
# Dataset Paths
# ==========================

dataset_base = "/content/merged_dataset/merged_dataset"

train_dir = os.path.join(dataset_base, "train")
test_dir = os.path.join(dataset_base, "test")

print(train_dir)
print(test_dir)
```

```python
# ==========================
# Data Generators
# ==========================

train_data_generator = ImageDataGenerator(
    rescale=1./255,
    rotation_range=15,
    width_shift_range=0.15,
    height_shift_range=0.15,
    shear_range=0.15,
    zoom_range=0.15,
    horizontal_flip=True
)

fer_training_data = train_data_generator.flow_from_directory(
    train_dir,
    target_size=(48,48),
    batch_size=64,
    color_mode='grayscale',
    class_mode='categorical'
)

test_data_generator = ImageDataGenerator(
    rescale=1./255
)

fer_test_data = test_data_generator.flow_from_directory(
    test_dir,
    target_size=(48,48),
    batch_size=64,
    color_mode='grayscale',
    class_mode='categorical'
)
```

```python
# ==========================
# Load Trained Model
# ==========================

model = load_model("/content/emotion_classification_model.keras")

print("Model Loaded Successfully")
model.summary()
```

```python
# ==========================
# Callbacks
# ==========================

early_stopping = EarlyStopping(
    monitor='val_accuracy',
    min_delta=0.00005,
    patience=11,
    verbose=1,
    restore_best_weights=True
)

lr_scheduler = ReduceLROnPlateau(
    monitor='val_accuracy',
    factor=0.5,
    patience=7,
    min_lr=1e-7,
    verbose=1
)

callbacks = [
    early_stopping,
    lr_scheduler
]
```

```python
# ==========================
# Continue Training
# ==========================

history = model.fit(
    fer_training_data,
    epochs=18,     # add 30 more epochs
    validation_data=fer_test_data,
    callbacks=callbacks
)
```

```python
# ==========================
# Evaluate
# ==========================

loss, accuracy = model.evaluate(
    fer_test_data,
    verbose=1
)

print("Accuracy:", accuracy * 100)
```

```python
# ==========================
# Save New Model
# ==========================

model.save("emotion_classification_model_v2.keras")

print("Saved Successfully")
```
